# **۱. معادله خط در رگرسیون:**
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در رگرسیون خطی مدل‌سازی خروجی (برچسب) بر اساس یک متغیر ورودی به صورت یک رابطه خطی است:
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
$ \hat{y} = wx + b $
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
 اگر قیمت خانه ($y$) را بر حسب متراژ ($x$) پیش‌بینی کنیم، $w$ و $b$ از نظر فیزیکی چه معنایی دارند؟
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
• $w$: به معنای میزان تغییر متوسط قیمت خانه وقتی متراژ یک متر افزایش می‌یابد. اگر $w$ برابر ۷ میلیون باشد یعنی با هر ۱ متر مربع افزایش، قیمت خانه به طور میانگین ۷ میلیون تومان زیاد می‌شود.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
• $b$: نشان‌دهنده قیمت پایه است. اگر متراژ صفر باشد (که در واقعیت معنایی ندارد)، مقدار $b$ قیمت خانه در آن نقطه فرضی است. این عدد صرفاً مقدار نقطه‌ای مدل و نقش “عرض از مبدأ” خط را دارد.
</div>
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
    
### مثال  
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اگر  
\( w = 10 \) و \( b = 50 \)  
باشند:
</div>
<div style="text-align:center;">
$ \hat{y} = 10 \times 100 + 50 = 1050 $
    (مثلاً واحد میلیون تومان)
</div>

----

## **۲. تابع هزینه: میانگین مربعات خطا (MSE)**
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
برای اندازه‌گیری خطای مدل از تابع هزینه‌ی زیر استفاده می‌شود:

<div style="text-align:center;">
$ J(w, b) = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2 $
</div>
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در این رابطه:

- **$N$**: تعداد نمونه‌ها  
- **$y_i$**: مقدار واقعی  
- **$\hat{y}_i$**: مقدار پیش‌بینی‌شده  
- هرچه مقدار **$J$ کوچک‌تر** باشد، مدل **دقیق‌تر** است.
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
    
# **چرا مربع؟**

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
- خطاهای بزرگ‌تر جریمه بیشتری دریافت کنند

## **۳. گرادیان کاهشی و نقش نرخ یادگیری $\eta$**

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در گرادیان کاهشی، وزن‌ها با استفاده از نرخ یادگیری $\eta$ به‌روزرسانی می‌شوند:

</div>
<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
$ w = w - \eta \frac{\partial J}{\partial w} $
</div>
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
    
### **وقتی $\eta$ خیلی بزرگ باشد:**

- وزن‌ها در هر گام **بیش از حد تغییر می‌کنند**.  
- به‌جای نزدیک شدن به نقطه‌ی کمینه، ممکن است از آن **عبور کنیم** و به سمت دیگر منحنی پرت شویم.  
- در موارد شدید، الگوریتم بین نواحی مختلف **نوسان می‌کند** یا کاملاً **واگرا می‌شود** (مقدار تابع هزینه دائماً بزرگ و بزرگ‌تر می‌شود).  
- در نتیجه مدل اصلاً **به جواب بهینه نمی‌رسد** و ممکن است عملکردش بسیار بد شود.

### **وقتی $\eta$ خیلی کوچک باشد:**

- هر گام به‌روزرسانی **بسیار ریز** خواهد بود.  
- الگوریتم به سمت کمینه حرکت می‌کند، ولی **سرعت همگرایی بسیار پایین** می‌شود.  
- برای رسیدن به نقطه‌ی مناسب، به **تعداد تکرارهای خیلی زیاد** نیاز است و آموزش مدل **زمان‌بر** می‌شود.  
- در عمل ممکن است قبل از رسیدن به جواب خوب، مجبور شویم آموزش را متوقف کنیم (به‌خاطر محدودیت زمان/منابع).

بنابراین انتخاب **Learning Rate مناسب** بسیار مهم است!
---

# **بخش ۲: روش های بهینه سازی و مفاهیم دسته بندی**

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
<b>1. روش حداقل مربعات (Least Squares) در رگرسیون خطی</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در مسئله رگرسیون خطی هدف این است که رابطه‌ای بین ورودی‌ها (ویژگی‌ها) و خروجی پیدا کنیم تا بتوانیم مقدار خروجی را پیش‌بینی کنیم. 
فرض می‌کنیم که بین متغیرهای ورودی و خروجی یک رابطه خطی وجود دارد. یعنی خروجی به صورت ترکیبی خطی از ویژگی‌ها مدل می‌شود.
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>\\(z = w^T x + b\\)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در این رابطه:
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
x بردار ویژگی‌ها است.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
w بردار وزن‌ها یا ضرایب مدل است که باید یاد گرفته شوند.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
b بایاس مدل است که باعث می‌شود مدل بتواند خط را جابجا کند.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در عمل مقدار پیش‌بینی شده مدل با مقدار واقعی دقیقاً برابر نیست، بنابراین بین مقدار واقعی و مقدار پیش‌بینی شده یک خطا وجود دارد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
روش حداقل مربعات سعی می‌کند وزن‌ها را طوری پیدا کند که مجموع مربعات خطاها حداقل شود.
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>\\(L = Σ (y_i - ŷ_i)^2\\)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در این فرمول y مقدار واقعی و ŷ مقدار پیش‌بینی شده توسط مدل است.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
با حل این مسئله به صورت تحلیلی می‌توان مستقیماً وزن‌های بهینه را محاسبه کرد.
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>\\( w = (X^T X)^{-1} X^T y \\)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در این رابطه X ماتریس داده‌ها است که در آن هر سطر یک نمونه و هر ستون یک ویژگی را نشان می‌دهد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
مزیت مهم روش حداقل مربعات این است که جواب دقیق و مستقیم به دست می‌دهد و نیازی به تکرارهای زیاد مثل روش گرادیان نزولی ندارد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اما یک مشکل مهم آن زمانی ظاهر می‌شود که حجم داده‌ها خیلی زیاد باشد. در این حالت باید ماتریس $ X^T X $ محاسبه و سپس معکوس آن گرفته شود که از نظر محاسباتی هزینه زیادی دارد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
به همین دلیل در مسائل یادگیری ماشین که داده‌ها بسیار بزرگ هستند معمولاً از روش‌های تکراری مثل گرادیان نزولی استفاده می‌شود چون به حافظه و محاسبات کمتری نیاز دارند.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در مجموع می‌توان گفت روش حداقل مربعات برای داده‌های کوچک یا متوسط بسیار مناسب است اما برای داده‌های بسیار بزرگ ممکن است کارایی خوبی نداشته باشد.
</div>



<br><br>


<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
<b>2. پرسپترون (Perceptron)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
پرسپترون یکی از ساده‌ترین و اولین مدل‌های شبکه عصبی مصنوعی است که برای مسائل دسته‌بندی استفاده می‌شود. 
ایده اصلی پرسپترون این است که ابتدا یک ترکیب خطی از ویژگی‌ها محاسبه شود و سپس با استفاده از یک تابع فعال‌سازی تصمیم گرفته شود که نمونه به کدام کلاس تعلق دارد.
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>\\(z = w^T x + b\\)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
پس از محاسبه مقدار z، یک تابع فعال‌سازی روی آن اعمال می‌شود تا خروجی نهایی به دست آید.
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>\\(y = f(z)\\)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در ساده‌ترین حالت از تابع پله‌ای استفاده می‌شود.
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
\\(
f(z)=
\begin{cases}
1 , & z \ge 0 \\\\
0 , & z < 0
\end{cases}
\\)
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
تفاوت مهم پرسپترون با رگرسیون خطی در همین تابع فعال‌سازی است. در رگرسیون خطی خروجی یک مقدار پیوسته است و برای پیش‌بینی مقادیر عددی استفاده می‌شود.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اما در پرسپترون خروجی به صورت گسسته است و مدل در نهایت تصمیم می‌گیرد که داده به کدام کلاس تعلق دارد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
به همین دلیل پرسپترون برای مسائل دسته‌بندی مناسب‌تر است، در حالی که رگرسیون خطی بیشتر برای مسائل پیش‌بینی عددی استفاده می‌شود.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
البته پرسپترون تنها زمانی به درستی کار می‌کند که داده‌ها خطی جداپذیر باشند، یعنی بتوان آن‌ها را با یک خط یا یک صفحه از هم جدا کرد.
</div>



<br><br>


<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
<b>3. دسته‌بندی خطی (Linear Classification) و مرز تصمیم</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در بسیاری از مسائل یادگیری ماشین هدف این است که داده‌ها به چند کلاس مختلف تقسیم شوند. 
در دسته‌بندی خطی فرض می‌کنیم که می‌توان کلاس‌ها را با استفاده از یک مرز خطی از هم جدا کرد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
مدل ابتدا مقدار زیر را محاسبه می‌کند:
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>\\( z = w x + b \\)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
این مقدار نشان می‌دهد که یک نمونه در کدام سمت مرز تصمیم قرار گرفته است.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
مرز تصمیم در واقع مجموعه نقاطی است که مقدار z برای آن‌ها صفر می‌شود.
</div>

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>\\( wx + b = 0 \\)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در فضای دو بعدی این مرز یک خط است که صفحه را به دو قسمت تقسیم می‌کند.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در فضای سه بعدی این مرز یک صفحه خواهد بود.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در فضاهای با ابعاد بالاتر نیز این مرز یک ابرصفحه یا Hyperplane نامیده می‌شود.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
مدل با توجه به علامت مقدار z تصمیم می‌گیرد که نمونه در کدام کلاس قرار بگیرد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اگر مقدار z مثبت باشد نمونه به کلاس مثبت تعلق می‌گیرد و اگر منفی باشد به کلاس دیگر تعلق دارد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
بنابراین مرز تصمیم در واقع خط یا صفحه‌ای است که مدل بر اساس آن بین کلاس‌ها تمایز ایجاد می‌کند.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
<b>4. ماتریس درهم‌ریختگی (Confusion Matrix)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
بعد از اینکه یک مدل دسته‌بندی ساخته شد، باید بررسی کنیم که مدل تا چه اندازه درست پیش‌بینی می‌کند. 
یکی از ابزارهای مهم برای ارزیابی مدل‌های دسته‌بندی، ماتریس درهم‌ریختگی است.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
این ماتریس نشان می‌دهد که مدل چه تعداد از نمونه‌ها را درست و چه تعداد را اشتباه پیش‌بینی کرده است.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
چهار حالت اصلی در این ماتریس وجود دارد:
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
True Positive : زمانی که مدل نمونه مثبت را درست به عنوان مثبت تشخیص دهد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
True Negative : زمانی که مدل نمونه منفی را درست به عنوان منفی تشخیص دهد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
False Positive : زمانی که مدل یک نمونه منفی را به اشتباه مثبت تشخیص دهد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
False Negative : زمانی که مدل یک نمونه مثبت را به اشتباه منفی تشخیص دهد.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
برای مثال فرض کنید یک مدل برای تشخیص ایمیل اسپم ساخته‌ایم.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اگر مدل یک ایمیل معمولی را به اشتباه اسپم تشخیص دهد، این حالت False Positive است.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اگر مدل یک ایمیل اسپم را تشخیص ندهد و آن را به عنوان ایمیل عادی در نظر بگیرد، این حالت False Negative است.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در بسیاری از کاربردها مثل تشخیص بیماری یا سیستم‌های امنیتی، False Negative می‌تواند بسیار خطرناک باشد چون باعث می‌شود یک مورد مهم شناسایی نشود.
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
به همین دلیل در تحلیل مدل‌ها معمولاً علاوه بر دقت کلی، مقادیری مثل Precision و Recall نیز بررسی می‌شوند تا عملکرد مدل بهتر ارزیابی شود.
</div>

توابع پیش‌بینی، هزینه و حلقه بهینه‌سازی
</div>


````python
def predict(X, w, b):
    # YOUR CODE HERE: Implement y_hat = wX + b
    y_hat = X.dot(w) + b
    return y_hat


def compute_cost(X, y, w, b):
    # YOUR CODE HERE: Implement Mean Squared Error (MSE)
    m = X.shape[0]
    y_hat = predict(X, w, b)
    cost = (1 / (2 * m)) * np.sum((y_hat - y) ** 2)
    return cost


def gradient_descent(X, y, w_init, b_init, alpha, epochs):
    w, b = w_init, b_init
    cost_history = []

    for i in range(epochs):
        y_hat = predict(X, w, b)
        # YOUR CODE HERE: Calculate gradients (dw, db) and update w, b
        m = X.shape[0]
        dw = (1 / m) * X.T.dot(y_hat - y)
        db = (1 / m) * np.sum(y_hat - y)

        w = w - alpha * dw
        b = b - alpha * db

        cost_history.append(compute_cost(X, y, w, b))

    return w, b, cost_history
````

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در ادامه، گرادیان کاهشی را اجرا کرده و نتیجه را بررسی می‌کنیم.
</div>


````python
# Initialize parameters
w_init = np.zeros((1, 1))
b_init = 0.0
alpha = 0.1
epochs = 1000

# Run gradient descent
w_gd, b_gd, cost_history = gradient_descent(x, y, w_init, b_init, alpha, epochs)

print("Estimated weight (w):", w_gd)
print("Estimated bias (b):", b_gd)
print("Final cost:", cost_history[-1])
````

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
برای مشاهده روند کاهش خطا، نمودار تابع هزینه بر حسب تکرارها را رسم می‌کنیم.
</div>


````python
plt.figure(figsize=(6, 4))
plt.plot(cost_history)
plt.xlabel("Epoch")
plt.ylabel("Cost (MSE)")
plt.title("Gradient Descent Cost over Epochs")
plt.grid(True)
plt.show()
````

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
همچنین داده‌ها و خط رگرسیون نهایی را ترسیم می‌کنیم.
</div>

````python
plt.figure(figsize=(6, 4))
plt.scatter(x, y, color="blue", label="Data")
x_line = np.linspace(0, 2, 100).reshape(-1, 1)
y_line = predict(x_line, w_gd, b_gd)
plt.plot(x_line, y_line, color="red", label="Linear Regression (GD)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression with Gradient Descent")
plt.legend()
plt.grid(True)
plt.show()
````

# **وظیفه ۲.۳: حل کمترین مربعات (Least Squares)**
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
به‌جای استفاده از درایورهای تکرارپذیر، وزن‌های بهینه را مستقیماً از طریق معادله نرمال برای داده‌های طبقه‌بندی زیر به دست آورید.
</div>


````python
from sklearn.datasets import make_blobs

# Generate classification data (Labels: -1 and 1)
X_p, y_p = make_blobs(n_samples=200, centers=2, n_features=2, random_state=42)
y_p = np.where(y_p == 0, -1, 1).reshape(-1, 1)

# YOUR CODE HERE: Add bias term (column of ones) to X_p
X_with_bias = np.c_[np.ones((X_p.shape[0], 1)), X_p]

# YOUR CODE HERE: Solve using Normal Equation: w = (X^T X)^-1 X^T y
w_lsp = np.linalg.inv(X_with_bias.T.dot(X_with_bias)).dot(X_with_bias.T).dot(y_p)

print("Least Squares Weights (with bias):")
print(w_lsp)
````

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
برای نمایش مرز تصمیم به‌دست‌آمده از روش کمترین مربعات، آن را روی داده‌ها رسم می‌کنیم.
</div>

````python
plt.figure(figsize=(6, 6))
plt.scatter(X_p[:, 0], X_p[:, 1], c=y_p.ravel(), cmap="bwr", edgecolors="k")

# Plot decision boundary: w0 + w1*x1 + w2*x2 = 0  →  x2 = -(w0 + w1*x1)/w2
w0, w1, w2 = w_lsp.ravel()
x1_vals = np.linspace(X_p[:, 0].min()-1, X_p[:, 0].max()+1, 100)
x2_vals = -(w0 + w1 * x1_vals) / w2
plt.plot(x1_vals, x2_vals, "g-", label="Least Squares Boundary")

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Least Squares Linear Classifier")
plt.legend()
plt.grid(True)
plt.show()
````

# **وظیفه ۳.۳: الگوریتم پرسپترون (Perceptron)**
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
در این قسمت باید الگوریتم پرسپترون را پیاده‌سازی کنید. هدف این است که در هر مرحله، در صورت تشخیص اشتباه، وزن‌ها را طبق قانون زیر به‌روزرسانی کنید:

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
$ w \leftarrow w + \eta (y_i \cdot x_i) $

</div>


````python
def perceptron_train(X, y, eta=0.1, epochs=10):
    # YOUR CODE HERE: Initialize weights (including bias) with zeros
    n_samples, n_features = X.shape
    w = np.zeros(n_features + 1)  # +1 for bias
    X_b = np.c_[np.ones((n_samples, 1)), X]

    for _ in range(epochs):
        for i in range(n_samples):
            # YOUR CODE HERE: Check the condition for misclassification
            # Hint: y_i * (w * x_i) <= 0
            if y[i] * np.dot(w, X_b[i]) <= 0:
                # YOUR CODE HERE: Update weights
                w = w + eta * y[i] * X_b[i]

    return w


w_perceptron = perceptron_train(X_p, y_p.ravel())
print("Perceptron Weights (with bias):")
print(w_perceptron)
````

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
برای مشاهده مرز تصمیم پرسپترون، آن را روی داده‌ها رسم می‌کنیم.
</div>

````python
plt.figure(figsize=(6, 6))
plt.scatter(X_p[:, 0], X_p[:, 1], c=y_p.ravel(), cmap="bwr", edgecolors="k")

# Decision boundary: w0 + w1*x1 + w2*x2 = 0
w0_p, w1_p, w2_p = w_perceptron
x1_vals = np.linspace(X_p[:, 0].min()-1, X_p[:, 0].max()+1, 100)
x2_vals = -(w0_p + w1_p * x1_vals) / w2_p
plt.plot(x1_vals, x2_vals, "k-", label="Perceptron Boundary")

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Perceptron Linear Classifier")
plt.legend()
plt.grid(True)
plt.show()
````

# **وظیفه ۴.۳: ارزیابی و ماتریس اغتشاش (Confusion Matrix)**
<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
مدل را بر روی داده‌های تست کرده و ماتریس اغتشاش را رسم کنید.
</div>


````python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

# YOUR CODE HERE: Generate predictions (sign of X_b * w)
X_p_b = np.c_[np.ones((X_p.shape[0], 1)), X_p]
y_pred = np.sign(X_p_b @ w_perceptron)

# YOUR CODE HERE: Compute confusion matrix
cm = confusion_matrix(y_p, y_pred)

sns.heatmap(cm, annot=True, fmt='d', cmap='Oranges', xticklabels=['-1', '1'], yticklabels=['-1', '1'])
plt.title("Perceptron Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

print(classification_report(y_p, y_pred))
````

به این روش مهندسی ویژگی (Feature Engineering) گفته می‌شود که با اضافه کردن ویژگی‌های جدید می‌توان روابط غیرخطی را با مدل خطی تقریب زد.

\hat{y} = w_1x_1 + w_2x_2 + \dots + w_nx_n + b

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
یک دانشجو این محاسبه را با استفاده از حلقه <code>for</code> به شکل زیر پیاده‌سازی کرده است:

<div dir="rtl" style="text-align:left; unicode-bidi:plaintext;">
    
```python
y_hat = 0
:for i in range(10)
   y_hat += w[i] * x[i]
y_hat += b




<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
این کد در واقع در حال محاسبه مجموع ضرب جزءبه‌جزء دو بردار وزن و ویژگی است

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
این همان تعریف دقیق ضرب داخلی (Dot Product) بین دو بردار است.

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">

<div dir="rtl" style="text-align: right; font-family: Vazirmatn, Tahoma; line-height: 1.9;">

به‌خصوص زمانی که تعداد ویژگی‌ها زیاد باشد یا دیتاست بزرگ باشد، محاسبهٔ ضرب داخلی بهترین روش برای پیاده‌سازی مدل رگرسیون خطی است.

<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>۹. معنی مقدار منفی برای R²</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اگر مقدار \(R^2\) برابر عدد منفی مثل ‎−0.15 باشد، این یعنی مدل حتی از یک مدل بسیار ساده که فقط میانگین خروجی‌ها را حدس می‌زند هم عملکرد بدتری داشته است.  
این اتفاق معمولاً زمانی رخ می‌دهد که مدل ساختار داده را یاد نگرفته یا داده‌ها نویزی و نامناسب هستند.  
بنابراین عدد منفی برای \(R^2\) نشانه ضعف شدید عملکرد مدل است.
</div>

---
<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>۱۰. اثر مقیاس ویژگی‌ها (Feature Scaling)</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اگر یک ویژگی در مقیاس خیلی کوچک (مثلاً 0.0012) و ویژگی دیگر در مقیاس خیلی بزرگ (مثلاً 1930) باشد، سطح تابع هزینه در فضای وزن‌ها شکل بیضی کشیده پیدا می‌کند.  
در چنین حالتی گرادیان نزولی مجبور است مسیر بسیار طولانی و دندانه‌دار طی کند و سرعت یادگیری پایین می‌آید.  
به همین دلیل استانداردسازی یا نرمال‌سازی ویژگی‌ها کمک می‌کند همه ویژگی‌ها در یک محدوده مشابه قرار بگیرند و یادگیری مدل سریع‌تر و پایدارتر انجام شود.
</div>

---
<div dir="rtl" style="text-align:center; unicode-bidi:plaintext;">
<b>۱۱. اختلاف عملکرد بین داده‌های آموزش و آزمون</b>
</div>

<div dir="rtl" style="text-align:right; unicode-bidi:plaintext;">
اگر مدل روی داده‌های آموزش امتیاز عالی مثل \(R^2 = 1.0\) کسب کند ولی روی داده‌های آزمون عملکرد ضعیفی داشته باشد، این نشانه واضح «بیش‌برازش» یا Overfitting است.  
یعنی مدل به‌جای یادگیری الگوی کلی، نویز و جزئیات خاص داده‌های آموزش را حفظ کرده است.  
به همین دلیل هنگام روبه‌رو شدن با داده‌های جدید (آزمون) عملکردش افت می‌کند و دقتش پایین می‌آید.
</div>
